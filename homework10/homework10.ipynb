```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from astropy import units as u
from astropy import constants as const
```

```python
# Import csv file as dataframe
df = pd.read_csv("../../course_assignments/homework9/solar_system.csv")

# Transpose data
print("Shape BEFORE transposing:", df.shape)
df_before = df.copy()

df = df.set_index("Attribute").T
print("Shape now:", df.shape)

df.index.name = "Planet"
print("Shape now:", df.shape)

df.reset_index(inplace=True)
print("Shape now:", df.shape)

df.columns.name = None
print("Shape now:", df.shape)

# Comparing Transposed Data
print("Shape BEFORE:", df_before.shape)
print("Shape AFTER:", df.shape)

rows_before, cols_before = df_before.shape
rows_after, cols_after = df.shape
print(f"Rows before: {rows_before}, Columns before: {cols_before}")
print(f"Rows after:  {rows_after}, Columns after:  {cols_after}")

# The lines of code above in part 4;
# Transposes the DataFrame properly and 
# Renames the index
# Resets it back to a column
# Removes column index name

# The shape is not (11, 20) because the original CSV has more than 20 attributes and includes non-planet rows.

print("New Column Names are", df.columns.tolist())

# Which Columns have units
unit_cols = []
non_unit_cols = []

for col in df.columns:
    if df[col].astype(str).str.contains(r"[A-Za-z]").any():
        unit_cols.append(col)
    else:
        non_unit_cols.append(col)

print("\nColumns WITH units:")
print(unit_cols)
print("Total:", len(unit_cols))

print("\nColumns WITHOUT units:")
print(non_unit_cols)
print("Total:", len(non_unit_cols))

print("\n=== TRANSPOSED DATAFRAME ===")
df
```

```python
df.dtypes

for col in df.columns:
    types_in_col = df[col].dropna().apply(type).unique()
    print(col, "→", types_in_col)

# The values are strings but they should be floats, NaN, and some are mixed types if they include the units
# Storing numeric values as strings is a problem for data analysis because numpy cannot treat them as values and cannot perform mathematical operations. Cleaning the data becomes harder as well.
```

```python
# Convert string into numbers
for col in df.columns:
    if col not in ["Planet", "Ring System?", "Global Magnetic Field?"]:
        df[col] = pd.to_numeric(df[col], errors="coerce")

df.dtypes

# The three data types are object, int64, and float64.
# Float64 is a 64-bit floating-point number and is used for real numbers. Int64 is a 64-bit integer, used when a column contains only whole numbers. Object is a generic Python object, usually a string column that is not interpreted as a number.
# Planet is still an object because it has the planets names not numbers. Ring system? and Global Magnetic Field? are still an object because they are yes or no, not numeric. 
```

```python
def attach_units(df, old_col, unit, new_col=None):
    if old_col not in df.columns:
        raise KeyError(f"Column '{old_col}' not found in DataFrame. Available columns: {df.columns.tolist()}")
    
    df[old_col] = [val * unit for val in df[old_col]]
    if new_col:
        df.rename(columns={old_col: new_col}, inplace=True)
    return df

df = attach_units(df, old_col="Mass (kg)", unit=1e24*u.kg, new_col="Mass (kg)")
print(df["Mass (kg)"])
```

```python
semi_major_axis = (df["Perihelion (10^6 km)"] + df["Aphelion (10^6 km)"]) / 2
# Attach as new column after Aphelion
aphelion_idx = df.columns.get_loc("Aphelion (10^6 km)")
df.insert(aphelion_idx + 1, "Semi-Major Axis (10^6 km)", semi_major_axis)

# Print all column names to confirm
print(df.columns)

# Print the new column
print(df["Semi-Major Axis (10^6 km)"] * u.km * 1e6)  # attach units
```

```python
# Convert to years
df["Orbital Period (yr)"] = df["Orbital Period (days)"] * u.day.to(u.year)  # multiply by factor

# Print all columns
print(df.columns)

# Print the full column
print(df["Orbital Period (yr)"])

earth_period = df.loc[df["Planet"] == "Earth", "Orbital Period (yr)"].values[0]
print(f"Earth's orbital period: {earth_period:.4f} yr")
```

```python
# AU is the average distance between sun and earth

# Print 1 AU
print("1 AU:", const.au)

# How many kilometers in 1 AU
print("1 AU in km:", const.au.to(u.km))

# Distance columns in km (or 10^6 km)
distance_cols = ["Diameter (km)", "Perihelion (10^6 km)", "Aphelion (10^6 km)", "Semi-Major Axis (10^6 km)"]

# Convert to AU using const.au
for col in distance_cols:
    if "10^6" in col:
        df[col.replace("(10^6 km)", "(AU)")] = df[col] * 1e6 / const.au.value
    else:
        df[col.replace("(km)", "(AU)")] = df[col] / const.au.value

# Print sample values for Earth
earth = df[df["Planet"] == "Earth"]
print("Sample distance values for Earth (in AU):")
for col in distance_cols:
    au_col = col.replace("(10^6 km)", "(AU)") if "10^6" in col else col.replace("(km)", "(AU)")
    print(f"{au_col}: {earth[au_col].values[0]:.6f} AU")
```
