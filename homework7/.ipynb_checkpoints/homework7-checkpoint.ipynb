```python
import numpy as np
import matplotlib.pyplot as plt

def make_sine_wave(x, A, w):
    y = A * np.sin(w * x)
    return y

x = np.linspace(0, 2 * np.pi, 1000)

amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]

fig, axes = plt.subplots(1, 3, figsize=(25, 5))

# --- First subplot: vary amplitude only ---
for A, w in zip(amplitude, frequency):
    y = make_sine_wave(x, A, 1) #set frequency to 1
    axes[0].plot(x, y, label=f"A={A}")
    axes[0].set_title("Varying Amplitude (ω=1)", fontsize=16)
    axes[0].set_xlabel("x", fontsize=14)
    axes[0].set_ylabel("y", fontsize=14)
    axes[0].legend(fontsize=12)
    axes[0].grid(True)

# --- Second subplot: Vary only the frequency ---
for A, w in zip(amplitude, frequency):
    y = make_sine_wave(x, 1, w) #set amplitude to 1
    axes[1].plot(x, y, label=f"w={w}")
    axes[1].set_title("Varying frequency (A=1)", fontsize=16)
    axes[1].set_xlabel("x", fontsize=14)
    axes[1].set_ylabel("y", fontsize=14)
    axes[1].legend(fontsize=12)
    axes[1].grid(True)

# --- Third subplot: Vary the amplitude and frequency together ---
for A, w in zip(amplitude, frequency):
    y = make_sine_wave(x, A, w)
    axes[2].plot(x, y, label=f"A={A}, w={w}")
    axes[2].set_title("Varying frequency and amplitude", fontsize=16)
    axes[2].set_xlabel("x", fontsize=14)
    axes[2].set_ylabel("y", fontsize=14)
    axes[2].legend(fontsize=12)
    axes[2].grid(True)

plt.tight_layout()
plt.show()
```
