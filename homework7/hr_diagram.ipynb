```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("../../course_assignments/homework7/star_data.txt",
skiprows=1)

print(data.shape) # 2642 rows

magnitude = data[:,0]
temperature = data[:,1]

# min and max magnitude
min_mag = np.min(magnitude)
max_mag = np.max(magnitude)

# Indices (locations) of those values
min_mag_index = np.argmin(magnitude)
max_mag_index = np.argmax(magnitude)

print("Minimum magnitude:", min_mag, "→ Star index:", min_mag_index)
print("Maximum magnitude:", max_mag, "→ Star index:", max_mag_index)

min_temp = np.min(temperature)
max_temp = np.max(temperature)
min_temp_index = np.argmin(temperature)
max_temp_index = np.argmax(temperature)

print("Minimum temperature:", min_temp, "→ Star index:", min_temp_index)
print("Maximum temperature:", max_temp, "→ Star index:", max_temp_index)

print(f"Brightest star index {min_mag_index} → magnitude = {min_mag}, temperature = {temperature[min_mag_index]}")
print(f"Dimmest star index {max_mag_index} → magnitude = {max_mag}, temperature = {temperature[max_mag_index]}")
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

plt.figure(figsize=(8, 6))
plt.scatter(log_temperature, log_magnitude, 
            s=10, alpha=0.6, color='blue', label='All Stars')
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
plt.scatter(np.log(temperature[min_mag_index]), np.log(magnitude[min_mag_index]),
            color='black', marker='*', s=300, label='Brightest Star')
plt.scatter(np.log(temperature[max_mag_index]), np.log(magnitude[max_mag_index]),
            color='purple', marker='*', s=300, label='Dimmest Star')
plt.scatter(np.log(temperature[max_temp_index]), np.log(magnitude[max_temp_index]),
            color='green', marker='*', s=300, label='Hottest Star')
plt.scatter(np.log(temperature[min_temp_index]), np.log(magnitude[min_temp_index]),
            color='orange', marker='*', s=300, label='Coldest Star')
plt.legend(fontsize=11)
plt.xlabel("Temperature (log K)", fontsize=12)
plt.ylabel("Magnitude", fontsize=12)
plt.title("Hertzsprung–Russell Diagram", fontsize=16)
```
