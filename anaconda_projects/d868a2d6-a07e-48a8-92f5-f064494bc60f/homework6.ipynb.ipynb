```python
"4"
arr = np.array([
[2, 3, 5],
[14, 6, 8],
[11, 13, 17],
[7, 10, 13]
])

def rows_with_primes(arr):
    def is_prime(n):
        if n < 2:
            return False
        for i in range(2, int(np.sqrt(n)) + 1):
            if n % i == 0:
                return False
        return True

    is_prime_vec = np.vectorize(is_prime)
    mask = np.any(is_prime_vec(arr), axis=1)
    return arr[mask]
print(rows_with_primes(arr))
```

```python
"5"
def checkerboard():
    board = np.zeros((8, 8), dtype=int)
    board[::2, ::2] = 1  
    board[1::2, 1::2] = 1  
    return board
print(checkerboard())
```

```python
def reverse_checkerboard():
    board = np.zeros((8, 8), dtype=int)
    board[1::2, ::2] = 1 
    board[::2, 1::2] = 1
    return board
print(reverse_checkerboard())
```

```python
np.random.seed(123)
stars = np.random.randint(500, 2000, (5, 5))
def secondDimmest(stars):
    sorted_rows = np.sort(stars, axis=1)
    return sorted_rows[:, -2]
print(secondDimmest(stars))
```
